```python
products = df[['StockCode','Description','UnitPrice']]

products = products.drop_duplicates(subset='StockCode')

products.reset_index(drop=True,inplace=True)

products.head()
```

```python
products = products[
    products["StockCode"].astype(str).str.match(r'^[A-Za-z0-9]+$')
]

products = products.dropna(subset=["Description"])

products.head()
```

```python
import random

categories = [
    "Home Decor",
    "Kitchen",
    "Electronics",
    "Stationery",
    "Gift",
    "Garden",
    "Furniture",
    "Accessories"
]

products["category"] = [
    random.choice(categories)
    for _ in range(len(products))
]
```

```python
products["supplier_id"] = [
    f"S{random.randint(1,100):03}"
    for _ in range(len(products))
]
```

```python
products["lead_time_days"] = [
    random.randint(2,14)
    for _ in range(len(products))
]
```

```python
products["cost_price"] = (
    products["UnitPrice"] * 0.65
).round(2)

products.rename(
    columns={
        "StockCode":"product_id",
        "Description":"product_name",
        "UnitPrice":"selling_price"
    },
    inplace=True
)
```

```python
products.head()
```

```python
products.to_json(
    "products.json",
    orient="records",
    indent=4
)
```

```python
inventory = products[["product_id"]].copy()
```

```python
warehouses = ["W001","W002","W003"]

inventory["warehouse_id"] = [
    random.choice(warehouses)
    for _ in range(len(inventory))
]
```

```python
inventory["current_stock"] = [
    random.randint(20,500)
    for _ in range(len(inventory))
]
```

```python
inventory["reorder_point"] = [
    random.randint(30,80)
    for _ in range(len(inventory))
]
```

```python
inventory["safety_stock"] = [
    random.randint(10,30)
    for _ in range(len(inventory))
]
```

```python
inventory["last_updated"] = pd.Timestamp.today()
```

```python
inventory.head()
```

```python
inventory.to_csv(
    "inventory.csv",
    index=False
)
```
